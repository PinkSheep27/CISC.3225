# Logistic Regression
Statistical method used for predictive analysis.

Models Relationships between dependent and independent variables by using a linear equation to observe the data.

- Simple -> Uses exactly one independent variable to predict the dependent variable.
- Multiple -> Uses two or more independent variables to predict the dependent variable.

## Equation
### y = wx + b
- y -> Predicted Value
- x -> Input Value
- w -> Weight/Slope 
    - How much y changes for a unit change in x
- b -> Bias/Y-Intercept
    - Value of y when x is 0.

## Line of Best Fit
The goal is to draw a straight line that best represents the overall trend of the data.

Algorithm calculates the Residuals(Errors).
- Vertical distance between an actual data point and the predicted point on the line.

### Residual = Actual y - Predicted y
y - y_hat

## Mean Squared Error (Cost Function)
The most common cost function for linear regression.

### (1/n) * Σ(y_i - ŷ_i)²
- Calculates the residual for every point
- Squares each residual
    - Makes sure negative and positive errors don't cancel each other out.
- Takes the average of all the squared errors.

Results in the lowest possible MSE.
- Solved using Gradient Descent.

## L.I.N.E.
L -> Linearity
- X and y must be linear (if data is curved this won't help)

I -> Independence 
- Observations must be independent of each other 

N -> Normality
- Residuals should be normally distributed 

E -> Equal Variance
- Spread should remain constant across all values of x. (Shouldn't become cone shaped)

# Data Munging
## What is it?
Process of taking raw data and cleaning, transforming, and formatting into a structure that a machine learning model can actually understand.

If you feed a model bad data, you'll get bad predictions.

## Why we need it?
ML models can't read text, can't handle missing blank spaces in a dataset, and they struggle when numbers are on drastically different scales.

## Techniques
### Handling Missing Values
Raw data has missing entries (NaN or Null).
- Drop
    - Delete the row containing the missing value entirely.

- Fill
    - Replace the missing blank with a calculated substitute, such as Mean, Median, and Mode column.

### Encoding Categorical Variables
Text categories must be converted into numbers.
- Label Encoding
    - Assigns a single integer to each category.
    - Best for categories that have a natural order.

- One-Hot Encoding
    - Creates a new separate binary column for every category.
    - Best for categories with no inherent order.

### Feature Scaling
If one feature has values from 1 to 10, and another has values from 10,000 to 100,000, the model might accidentally assume the large numbers are more "important".

- Min-Max Scaling
    - Squeezes all values to be exactly between 0 and 1.

- Standardization
    - Centers the data around a mean of 0, with a standard deviation of 1.

## Train/Test Split
### Breakdown
- Training Set
    - 70% - 80% of the data.
    - Model uses this data to learn.
    - Looks at the features (X) and the targets (y) and figures out the math that connects them.

- Testing Set
    - 20% - 30% of the data.
    - Model has never seen this data.
    - Hide the y and ask the model to make predictions.
    - Compare its predictions to the hidden true answers to grade its accuracy.

### Why it's necessary
It's the only way to measure a model's ability to make accurate predictions in the real world on data it hasn't encountered yet.

Problems:
- Memorization
    - Model learns the training set too perfectly, including all the random noise and outliers.

- Underfitting
    - Model is too basic to capture the patterns in the data, resulting in poor scores for both Training and Testing Sets.

## Model Training
### Gradient Descent
Basically an optimization algorithm.

Step 1 (Initialization)
- Starts by assigning random weights to all features.

Step 2 (Prediction)
- Model makes a prediction for every single data point in the training set.

Step 3 (Error)
- Model compares its predictions against the actual true answers. 
- It calculates the overall error using a Loss Function.

Step 4 (Optimization)
- Gradient Descent calculates which direction it needs to change the weights to make the loss smaller.

Step 5 (Weight Update)
- Size of this tweak is controlled by the Learning Rate.
- If learning rate too large, the model might overshoot the optimal answer.
- If learning rate too small, training will take forever.

Step 6 (Repeat Process)
- Repeat steps 2-5 over and over.
- Each full pass over the training data. (Epoch)
- Model keeps updating the weights until the loss stops decreasing. (Convergence)

### Hyperparameters vs. Parameters
Parameters
- These are the variables the model learns on its own during training.

Hyperparameters
- Variables you set before training begins to control how the model learns.

## Predictions
### What is a Prediction?
Prediction phase is when you pass brand-new, unseen features (X) into this fixed model, and it outputs its best guess (y_hat).

### How the Mechanics Change by Model
Formula-Based Models
- Model takes new input features and plugs them directly into the algebraic equation it learned during training.
    - Making predictions is way faster with these models because it's just running a fixed equation.

Instance Based Models
- When a new data point comes in, it calculates the Euclidean distance between the new point and every single point in its training database.
- Finds the k closest points and takes a majority vote to make its prediction.

### Predictions vs. Probability Predictions
Discrete Predictions (model.predict())
- Gives you a definitive answer.
- Model looks at all possibilities and outputs a single label.

Probability Predictions (model.predict_porba())
- Answer consisting of raw probability distribution across every single possible category.
- All probabilities will add up to exactly 1.0 .

### Confidence vs. Accuracy
Confidence does not equal correctness.

A model can have an output of 99.8% confidence and still be completely wrong if it encounters an edge case or pattern it wasn't exposed to during training.

A massive part of error analysis involves isolating the most confident wrong predictions to understand exactly where the model's logic breaks down.

## Probability Scores
### How the Model Calculates Probabilities
To turn the raw data into clean, meaningful probabilities, the model passes them through an activation function.

- Signmoid Function
    - Used for Binary Classification (Yes or No)

- Softmax Function
    - Used for Multiclass Classification (Predicting between 3 or more classes)

### Two Mathematical Rules
- Bounded Range
    - Every single individual score is strictly bounded between 0.0 and 1.0.

- Sum to 1.0
    - All the individujal probabilities should add upto 1.0

### Code Implementation
In libraries like Scikit-Learn, you extract these numbers using the predict_proba() method.

For a multiclass problem like the digits dataset, calling model.predict_proba([X_test[i]]) returns an array of 10 numbers.

The discrete prediction you get from model.predict() is simply the result of an argmax operation—meaning the computer looks at that array of 10 probabilities and returns the index of whichever value is the highest.

## Confidence vs. Accuracy
Confidence
- How sure the model claims it is. 
- It is an internal probability score between 0% and 100%

Accuracy
- Whether the model is actually right.
- It is an external check comparing the model's guess against the true, real-world label.

## Accuracy Score
### What is it?
Out of all the predictions the model made, what percentage did it get right?

### Formula
Number of Correct Predictions / Total Number of Predictions

### Code
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, y_pred)

### Trap
It's a good metric only if your dataset is balanced.
If it's imbalanced, accuracy becomes completely useless.

## Error Analysis

Involves digging into the failures to find an actionable pattern.

- Filter the test data to find only the rows where the model's prediction doesn't match the actual ground truth.
- EX
    - wrong_indices = np.where(y_pred != y_test)[0]

Visualizing Error Patterns via Confusion Matrices
- A global way to perform error analysis is visually inspecting a confusion matrix.
- Rather than checking individual rows, you check the off-diagonal cells to look for disproportionately high numbers.

## Confusion Matrices
### What are they?
- Tabular layout that visualizes the performance of a classification algorithm.
- Shows you exactly how it is wrong by mapping actual ground-truth labels directly against the model's predicted labels.

### Binary Classification
True Positive (TP)
- The model predicted a positive outcome, and it is actually positive.

True Negative (TN)
- The model predicted a negative outcome, and it is actually negative.

False Positive (FP)
- The model predicted a positive outcome, but it is actually negative.

False Negative (FN)
- The model predicted a negative outcome, but it is actually positive.

### Multiclass Matrix 
Matrix Orientation
- Scikit-learn configures True Labels on the Rows (Y-axis) and Predicted Labels on the Columns (X-axis).

Diagonal
- The diagonal line from the top-left corner to the bottom-right corner represents all the correct predictions.

### Code
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

cm = confusion_matrix(y_test, y_pred)

disp = ConfusionMatrixDisplay(confusion_matrix=cm)

disp.plot(cmap="Blues")
plt.show()

### Derived Metrics
Accuracy
- All Diagonal Cells / Sum of All Cells in the Entire Matrix

Precision
- TP/ TP + FP

Recall
- TP / TP + FN

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, ConfusionMatrixDisplay

# ==========================================
# TOPIC 1: LOADING DATASETS
# ==========================================
print("--- 1. Loading Datasets ---")
# We load the pre-packaged digits dataset from scikit-learn
digits = load_digits()
X_raw = digits.data    # The features (pixel intensities)
y = digits.target      # The target labels (numbers 0-9)
print(f"Loaded {X_raw.shape[0]} images, each with {X_raw.shape[1]} pixels.\n")


# ==========================================
# TOPIC 2: DATA MUNGING
# ==========================================
print("--- 2. Data Munging ---")
# Machine learning models work best when numbers are on a similar scale.
# We "munge" (preprocess) the data by standardizing it so pixel values aren't too large.
scaler = StandardScaler()
X_clean = scaler.fit_transform(X_raw)
print(f"First pixel value before scaling: {X_raw[0][1]}")
print(f"First pixel value after scaling:  {X_clean[0][1]:.2f}\n")


# ==========================================
# TOPIC 3: TRAIN/TEST SPLIT
# ==========================================
print("--- 3. Train/Test Split ---")
# We hide 25% of the data from the model so we can test its true generalization later.
X_train, X_test, y_train, y_test = train_test_split(
    X_clean, y, test_size=0.25, random_state=42, stratify=y
)
print(f"Training on {X_train.shape[0]} samples. Testing on {X_test.shape[0]} samples.\n")


# ==========================================
# TOPIC 4: LOGISTIC REGRESSION (The Algorithm)
# TOPIC 5: MODEL TRAINING      (The Process)
# ==========================================
print("--- 4 & 5. Logistic Regression & Model Training ---")
# We initialize the Logistic Regression algorithm (Topic 4)
model = LogisticRegression(max_iter=1000, random_state=42)

# We TRAIN the model using the .fit() method (Topic 5)
# The model is learning the weights/math connecting the pixels to the numbers.
model.fit(X_train, y_train)
print("Model training complete!\n")


# ==========================================
# TOPIC 6: PREDICTIONS
# ==========================================
print("--- 6. Predictions ---")
# We pass the unseen Test data into the fixed model to get "hard" definitive predictions.
y_pred = model.predict(X_test)
print(f"First 5 true labels:       {y_test[:5]}")
print(f"First 5 model predictions: {y_pred[:5]}\n")


# ==========================================
# TOPIC 7: PROBABILITIES SCORES
# ==========================================
print("--- 7. Probability Scores ---")
# Instead of hard predictions, we ask the model for its raw Softmax probability distribution.
y_probs = model.predict_proba(X_test)
print("Probability distribution for the very first test image (0 through 9):")
# Showing the probabilities rounded to 4 decimal places
print(np.round(y_probs[0], 4), "\n")


# ==========================================
# TOPIC 8: CONFIDENCE VS. ACCURACY
# ==========================================
print("--- 8. Confidence vs. Accuracy ---")
# Confidence is how sure the model is (the highest probability score for each image).
confidence_scores = y_probs.max(axis=1)

# Let's find an example where the model was highly confident, but ACTUALLY WRONG.
wrong_mask = (y_pred != y_test) # Boolean mask of mistakes
if wrong_mask.any():
    # Find the index of the highest confidence score among the mistakes
    confident_but_wrong_idx = np.where(wrong_mask)[0][np.argmax(confidence_scores[wrong_mask])]
    print(f"DANGER! The model predicted a '{y_pred[confident_but_wrong_idx]}'")
    print(f"It was {confidence_scores[confident_but_wrong_idx] * 100:.2f}% Confident.")
    print(f"But the True Label was actually '{y_test[confident_but_wrong_idx]}'.")
    print("This proves Confidence does NOT always equal Accuracy!\n")


# ==========================================
# TOPIC 9: ACCURACY SCORE
# ==========================================
print("--- 9. Accuracy Score ---")
# We grade the model by dividing correct predictions by total predictions.
accuracy = accuracy_score(y_test, y_pred)
print(f"Overall Model Accuracy: {accuracy * 100:.2f}%\n")


# ==========================================
# TOPIC 10: ERROR ANALYSIS
# ==========================================
print("--- 10. Error Analysis ---")
# We isolate exactly where the model failed to find blind spots.
wrong_indices = np.where(y_pred != y_test)[0]
print(f"The model made {len(wrong_indices)} total mistakes.")
print(f"The indices of the first 5 mistakes are: {wrong_indices[:5]}\n")
# In a real workflow, we would use matplotlib to visually inspect these exact images!


# ==========================================
# TOPIC 11: CONFUSION MATRICES
# ==========================================
print("--- 11. Confusion Matrices ---")
# We generate a matrix to see exactly WHICH digits are confusing the model.
cm = confusion_matrix(y_test, y_pred)
print("Raw Confusion Matrix Data:")
print(cm)
print("\n(A plot window will now open. Close the window to end the script...)")

# Plotting the visual matrix
disp = ConfusionMatrixDisplay(confusion_matrix=cm)
disp.plot(cmap="Blues", colorbar=False)
plt.title("Confusion Matrix: Reality vs. Model Guesses")
plt.xlabel("Predicted Label (Model's Guess)")
plt.ylabel("True Label (Reality)")
plt.show()
```
